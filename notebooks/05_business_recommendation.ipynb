# BR-01 | Business Recommendations

## 1. Business Objective

RFM 고객 세분화 분석 결과를 바탕으로 고객 그룹별 구매 행동과 과거 매출 기여도를 고려한 차별화된 마케팅 전략을 제안한다.

총 **5,878명**의 고객을 7개 세그먼트로 분류한 결과, Champion과 Loyal Customer는 전체 고객의 **32.36%**를 차지하면서 전체 RFM 매출의 **75.82%**를 기여하였다.

따라서 모든 고객에게 동일한 마케팅 자원을 투입하기보다, 고객의 구매 활동과 과거 매출 기여도에 따라 고객 유지(Retention), 구매 활성화(Engagement), 재활성화(Reactivation) 전략을 구분할 필요가 있다.

## 2. Segment-Based Marketing Strategies

| Customer Segment | Recommended Strategy | Business Objective |
|---|---|---|
| **Champion** | VIP 혜택, 신제품 우선 구매 기회, 프리미엄 고객 서비스 제공 | 핵심 고가치 고객 유지 및 장기적 관계 강화 |
| **Loyal Customer** | 로열티 프로그램, 개인화 상품 추천, 교차 판매(Cross-selling) | 반복 구매 확대 및 고객 충성도 강화 |
| **Potential Loyalist** | 맞춤형 프로모션, 구매 주기별 이메일 캠페인 | 구매 빈도 증가 및 Loyal Customer 전환 유도 |
| **Need Attention** | 구매 리마인더, 기간 한정 혜택, 과거 구매 기반 개인화 캠페인 | 최근 구매 활동 감소 고객의 재구매 유도 |
| **At Risk** | 과거 구매 금액을 고려한 선별적 Win-back 캠페인 및 고객 피드백 조사 | 과거 구매 활동이 활발했던 고객의 재활성화 |
| **Lost Customer** | 과거 구매 가치와 캠페인 비용을 고려한 선택적 재활성화 | 재활성화 가능성 검증 및 마케팅 비용 효율화 |
| **Others** | 추가 구매 행동 분석, 세그먼트 특성 파악 및 타겟팅 개선 | 고객 특성 이해 및 향후 세분화 정교화 |

## 3. Key Insight

- **Customer Retention:** Champion과 Loyal Customer의 매출 기여도가 높으므로 핵심 고객 유지 전략을 우선 검토한다.
- **Growth Opportunity:** Potential Loyalist의 구매 빈도와 구매 금액을 높이기 위한 육성 캠페인을 검토한다.
- **Selective Reactivation:** Need Attention과 At Risk는 과거 구매 활동을 고려해 재활성화 우선순위를 설정한다.
- **Marketing Efficiency:** Lost Customer와 Others는 고객별 가치와 캠페인 비용을 고려해 차별화된 접근을 적용한다.

## 4. Analytical Limitation

본 전략은 과거 구매 이력에 기반한 RFM 분석 결과를 활용한 비즈니스 제안이다.

고객 세그먼트는 실제 이탈 여부나 미래 구매 가능성을 예측한 결과가 아니며, 제안된 마케팅 전략의 효과는 향후 A/B Testing 및 캠페인 KPI 분석을 통해 검증해야 한다.

# BR-02 | Strategic Priorities

RFM 고객 세분화 결과를 바탕으로 고객의 과거 매출 기여도, 구매 빈도 및 최근 구매 활동을 고려하여 마케팅 실행 우선순위를 제안한다.

| Priority | Target Segment | Strategic Focus | Recommended KPI |
|---|---|---|---|
| **P1 — Retention** | Champion, Loyal Customer | 핵심 고객 유지 및 반복 구매 강화 | Repeat Purchase Rate, Retention Rate |
| **P2 — Growth** | Potential Loyalist | 구매 빈도 증가 및 충성 고객 전환 | Purchase Frequency, Segment Conversion Rate |
| **P3 — Reactivation** | Need Attention, At Risk | 과거 구매 활동과 고객 가치를 고려한 선별적 재활성화 | Reactivation Rate, Campaign Conversion Rate |
| **P4 — Selective Engagement** | Lost Customer, Others | 고객별 가치 검토 및 비용 효율적인 타겟팅 | Cost per Reactivated Customer, Campaign ROI |

위 우선순위는 RFM 분석 결과에 기반한 전략적 제안이며, 실제 캠페인 성과나 고객별 수익성을 검증한 결과는 아니다.

# BR-03 | Recommendation Summary

## Key Recommendations

**1. Prioritize High-Value Customer Retention**

Champion과 Loyal Customer는 전체 RFM 고객 5,878명 중 **1,902명(32.36%)**이지만, 전체 RFM 매출 **£17,374,804.27의 75.82%**를 기여하였다.

따라서 신규 고객 확보에만 집중하기보다 기존 핵심 고객의 유지와 반복 구매 강화를 우선 검토할 필요가 있다.

**2. Develop Potential Loyalists**

Potential Loyalist는 **1,045명(17.78%)**으로, 구매 빈도 증가와 고객 충성도 강화를 위한 육성 전략의 주요 대상이다.

개인화 추천, 구매 리마인더 및 타겟 프로모션을 통해 반복 구매를 유도하는 방안을 검토한다.

**3. Apply Selective Reactivation Strategies**

Need Attention과 At Risk 고객은 총 **745명(12.67%)**으로, 최근 구매 활동과 과거 구매 이력을 함께 고려한 재활성화 전략이 필요하다.

고객별 과거 매출 기여도와 캠페인 비용을 비교하여 우선순위를 설정한다.

**4. Allocate Marketing Resources Based on Evidence**

고객 수만으로 마케팅 예산을 배분하기보다 고객별 구매 활동, 과거 매출 기여도 및 캠페인 성과를 함께 고려해야 한다.

실제 예산 배분과 캠페인 확대 여부는 추가적인 실험과 성과 측정을 통해 결정한다.

# BR-04 | Limitations

본 분석은 비즈니스 의사결정을 지원하지만 다음과 같은 한계가 존재한다.

- **Historical Transaction Data:** 과거 거래 데이터에 기반한 분석으로, 미래 구매 행동이나 실제 고객 이탈 여부를 예측하지 않는다.
- **Limited Customer Attributes:** 연령, 성별, 고객 선호도 등 인구통계 및 행동 관련 추가 정보가 포함되지 않았다.
- **No Predictive CLV:** 고객별 과거 누적 매출과 평균 매출을 분석했으나, 미래 Customer Lifetime Value(CLV)를 예측하지 않았다.
- **Rule-Based Segmentation:** RFM 점수와 사전에 정의한 규칙을 이용한 세분화로, 모든 고객 행동 특성을 설명하지 못할 수 있다.
- **External Factors:** 계절성, 프로모션, 시장 환경 및 경제적 요인의 영향을 별도로 검증하지 않았다.
- **Unvalidated Campaign Outcomes:** 제안된 전략의 실제 매출 증대 효과, 고객 유지 효과 및 ROI는 아직 검증되지 않았다.

# BR-05 | Future Work

향후 분석의 정확성과 마케팅 전략의 실행 가능성을 높이기 위해 다음 과제를 검토한다.

1. **Cohort & Retention Analysis:** 고객의 최초 구매 시점에 따른 재구매 및 유지 패턴 분석
2. **Predictive CLV Modeling:** 구매 이력을 활용한 미래 고객 가치 예측 모델 개발
3. **Customer Segmentation Enhancement:** K-Means 등 군집화 기법과 기존 규칙 기반 RFM 세분화 결과 비교
4. **Campaign A/B Testing:** 고객 세그먼트별 캠페인 효과 및 전환율 검증
5. **Marketing Performance Measurement:** Retention Rate, Reactivation Rate, Campaign ROI 등 핵심 KPI 모니터링

## Final Conclusion

본 프로젝트는 거래 데이터 분석을 통해 고객별 구매 행동과 매출 기여도의 차이를 확인하고, 이를 고객 세분화 및 마케팅 전략 제안으로 연결하였다.

특히 **핵심 고객 유지, 성장 잠재 고객 육성, 선별적 재활성화**를 주요 전략 방향으로 제안한다.

향후에는 실제 캠페인 데이터를 활용한 성과 검증을 통해 분석 결과를 실행 가능한 마케팅 의사결정 체계로 발전시킬 수 있다.
